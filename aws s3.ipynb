```python
import boto3
```

```python
AWS_REGION='us-east-1'
SECRET_KEY=''
ACCESS_KEY= ''
```

```python
client=boto3.client("s3", verify=False,
                   aws_secret_access_key=SECRET_KEY,
                   aws_access_key_id=ACCESS_KEY,
                   region_name=AWS_REGION)
```

```python
client
```

```python
# List Buckets

response=client.list_buckets()

print('Listing Buckets...')
for bucket in response['Buckets']:
    print(f'-- {bucket["Name"]} --')
```

```python
print(response)
```

```python
# create a bucket
bucket_name='prateek-1811-pv'
#location={"LocationConstraint":AWS_REGION}
response=client.create_bucket(Bucket=bucket_name)
                             #CreateBucketConfiguration=location)
print(f'Amazon S3 Bucket Created: {bucket_name}')
```

```python
# upload files to bucket

def upload_files(file_name, bucket, object_name=None, args=None):
    if object_name==None:
        object_name=file_name
    client.upload_file(file_name, bucket, object_name, ExtraArgs=args)
    print(f'{file_name} has been uploaded to bucket {bucket}')
```

```python
upload_files('C:\\Users\\pvashishtha\\Desktop\\hello.txt', 'prateek-1811-pv')
upload_files('C:\\Users\\pvashishtha\\Desktop\\index.html', 'prateek-1811-pv')
```

```python
# Download a file

s3_resource=boto3.resource("s3", verify=False,
                          aws_secret_access_key=SECRET_KEY,
                          aws_access_key_id=ACCESS_KEY,
                          region_name=AWS_REGION)
s3_object=s3_resource.Object('prateek-1811-pv', 'index.html')
s3_object.download_file('mypage.html')
print('File Downloaded!!!')
```
